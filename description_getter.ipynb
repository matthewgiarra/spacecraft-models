```python
# Imports
import json
from pathlib import Path
import requests
import os

# Define the base URL for the spacecraft info json files
base_url = 'https://eyes.nasa.gov/apps/solar-system/descriptions'

# Define path to the input JSON file
input_file = "spacecraft.json"

# Define output file
out_file = "spacecraft_with_descriptions.json"

# Load the source JSON file (spacecraft.json)
with open(input_file, "r") as f:
   file_data = json.load(f)

# Store the Spacecraft field as a list of dicts
Spacecraft = file_data['Spacecraft']

# For each spacecraft in Spacecraft, read the path to the EOTSS page (from the spacecraft[URL] field of spacecraft)
for spacecraft in Spacecraft:
    
    # Status
    print("Processing %s" % spacecraft["DisplayName"])
    
    # If the description is empty, retrieve it
    if not spacecraft["Description"]:
    
        # Get the trailing string, which should be the spacecraft ID
        name = Path(spacecraft['URL']).name.split('?')[0] + '.json'

        # Concatenate the base URL, spacecraft ID, and ".json" to create the path to the EOTSS description file
        json_url = base_url + "/" + name

        # Download the description json file at the URL you just constructed
        try:
            # Send a GET request to the URL
            response = requests.get(json_url, timeout=10)

            # Raise an exception for HTTP errors (e.g., 404, 500)
            response.raise_for_status()

            # Parse the response payload directly into a Python dictionary or list
            eotss_data = response.json()

        except requests.exceptions.RequestException as e:
            print(f"An error occurred: {e}")

        # Concatenate the ["decription"]["blurb"] and ["description"]["more"] fields to create 
        # the description 
        blurb = ''
        more = ''
        if "blurb" in eotss_data['description']:
            blurb = eotss_data['description']["blurb"]
        if "more" in eotss_data['description']:
            more = eotss_data["description"]["more"]    

        # Add description to the current spacecraft
        spacecraft['Description'] = blurb + "\n\n" + more
        if len(spacecraft['Description']) == 0:
            print("No description for %s" % spacecraft['Name'])

# Top-level json dict for writing to file
out_data = {"Spacecraft" : Spacecraft}

# Write json to file
with open(out_file, "w") as f:
    json.dump(out_data, f)

if os.path.isfile(out_file):
    print("Wrote %s" % out_file)
else:
    print("%s not saved!" % out_file)
```
